```python
# Make a rectangular image square by adding background padding only
# - Keeps the original pixels unchanged (no resize/crop)
# - Background color = median color of a 10-px border around the image

from PIL import Image
import numpy as np
from pathlib import Path

# === Set your source image ===
src_path = Path("../figures/tfus_fmri_final.png")   # <- your image
out_path = src_path.with_name(src_path.stem + "_square.png")

im = Image.open(src_path).convert("RGB")
w, h = im.size
size = max(w, h)                      # square canvas
k = 10                                # border width for color sampling

# sample border pixels to choose a good background color
arr = np.array(im)
border = np.concatenate([
    arr[:k, :, :].reshape(-1, 3),     # top
    arr[-k:, :, :].reshape(-1, 3),    # bottom
    arr[:, :k, :].reshape(-1, 3),     # left
    arr[:, -k:, :].reshape(-1, 3),    # right
], axis=0)
bg_color = tuple(np.median(border, axis=0).astype(np.uint8))

# create square canvas and paste centered
canvas = Image.new("RGB", (size, size), bg_color)
offset = ((size - w) // 2, (size - h) // 2)
canvas.paste(im, offset)

# save (300 dpi for print)
canvas.save(out_path, dpi=(300, 300))
print("Saved:", out_path)
print("Original (w,h):", (w, h), "New:", canvas.size, "BG:", bg_color)
```
